```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((200, 600), dtype=np.uint8)
cv2.putText(img, "dhashvanth", (30, 140), cv2.FONT_HERSHEY_SIMPLEX,
            3, 255, 5, cv2.LINE_AA)

# Step 5: Structuring element (kernel)
kernel = np.ones((5, 5), np.uint8)

# Step 6: Erosion shrinks the white text inward
eroded = cv2.erode(img, kernel, iterations=1)

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(eroded, cmap="gray")
plt.title("Eroded Image")
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((200, 600), dtype=np.uint8)
cv2.putText(img, "dhashvanth", (30, 140), cv2.FONT_HERSHEY_SIMPLEX,
            3, 255, 5, cv2.LINE_AA)

# Same kernel as erosion
kernel = np.ones((5, 5), np.uint8)

# Step 7: Dilation expands the white text outward
dilated = cv2.dilate(img, kernel, iterations=1)

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(dilated, cmap="gray")
plt.title("Dilated Image")
plt.axis("off")
plt.show()
```
